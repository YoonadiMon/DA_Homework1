3. List the distinct values of `payment_type` with the number of trips for each. Then convert the column to a category type and print its categories.

4. What are the mean and the median of `trip_distance`? Which is larger, and what does that suggest?

```python
mean = df.trip_distance.mean() # np.float64(8.484036652768351)
median = df.trip_distance.median() # np.float64(2.13)

print("Mean:", mean, ", Median:", median)
print("Mean (average) is larger than Median, suggesting that data is right-skewed, meaning there's more data on the left/smaller end.")
```

6. How many trips are longer than 10 miles?

```python
long_trip = df.trip_distance > 10
long_trip.sum()
(df.trip_distance > 10).sum()
```

7. How many trips were paid by credit card (`payment_type` 1) and carried two or more passengers?

```python
card = df.payment_type == 1 
passengers = df.passenger_count >= 2
(card & df.passenger_count >= 2).sum() # 0, ans in class -> WHY?
(card & passengers).sum() # 4465
```

8. Add a column `fare_per_mile` (`fare_amount` divided by `trip_distance`) for the trips with a distance above 0. What is its mean?

```python
moved_trips = df.trip_distance > 0
valid_trips = df.loc[moved_trips].copy()
valid_trips["fare_per_mile"] = valid_trips.fare_amount / valid_trips.trip_distance
df["fare_per_mile"] = valid_trips.fare_per_mile
df.fare_per_mile.mean()
```

9. Add a column `long_trip` that is `"Yes"` when `trip_distance` is 5 miles or more and `"No"` otherwise. How many trips are `"Yes"`?

```python
df["long_trip"] = np.where(df.trip_distance >= 5, "Yes", "No")
(df.long_trip == "Yes").sum()
```

10. Find the mean of `passenger_count`, then the mean with missing values included (`skipna=False`). If they differ, explain why in a markdown cell.

```python
print(df.passenger_count.mean())
print(df.passenger_count.mean(skipna=False))
```

- Without `skipna=False`, mean() skips NaN values by default. Not skipping NaN with `skipna=False` means the result will be NaN once there is a NaN or missing value in the data since the mean cannot be calculated.
